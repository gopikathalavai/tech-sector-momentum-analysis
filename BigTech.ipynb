```python
import yfinance as yf
import pandas as pd
import numpy as np
import os

stocks = ["AAPL", "MSFT", "NVDA", "AMZN", "META"]

data = yf.download(
    stocks,
    start="2024-01-01",
    end="2026-05-01"
)

close = data["Close"]

# Daily returns
returns = close.pct_change()

# Rolling volatility
volatility = returns.rolling(20).std()

# Create data folder
os.makedirs("data", exist_ok=True)

# Save files
close.to_csv("data/stock_prices.csv")
returns.to_csv("data/daily_returns.csv")
volatility.to_csv("data/rolling_volatility.csv")

print("Files saved successfully.")
```
